# Data Loading Testing Notebook

```python
from dataset import SoccerNetGSDataset
from torchvision import transforms
from torchvision.transforms import Resize

dataset = SoccerNetGSDataset("../../data/SoccerNetGS/train", transform=transforms.Compose([transforms.ToTensor(), Resize((100, 40))]))
```

```python
from sampler import SoccerNetGSBatchSampler
sampler = SoccerNetGSBatchSampler(dataset.labels, 2, 2)
```

```python
from torch.utils.data import DataLoader
loader = DataLoader(dataset, batch_sampler=sampler)

# dataset testing
print(dataset[0])

# sampler testing
print(f"Number of batches: {len(sampler)}")

# loader testing
for data in loader:
    print(data)
    break
```
